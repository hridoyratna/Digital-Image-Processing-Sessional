```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('fundas3.jpg',0)
```

```python
plt.figure(figsize=(8,5), dpi=150)
plt.imshow(img, cmap='gray')
plt.axis('off')
plt.show()

```

```python
# kernel 1
kernel = np.array([[0, 1, 0],
                   [1, -4, 1],
                   [0, 1, 0]])

```

```python
LaplacianImage = cv2.filter2D(src=img,
                              ddepth=-1,
                              kernel=kernel)

plt.figure(figsize=(8,5), dpi=150)
plt.imshow(LaplacianImage, cmap='gray')
plt.axis('off')
plt.show()

```

```python
c = -1


img_float = img.astype(np.float32)
LaplacianImage_float = LaplacianImage.astype(np.float32)

g = img_float + c * LaplacianImage_float


g = np.clip(g, 0, 255).astype(np.uint8)

plt.figure(figsize=(8,5), dpi=150)
plt.imshow(g, cmap='gray')
plt.axis('off')
plt.show()
```
